```python
##创建一个10*10的array A，并用1-100的随机值填充。
import numpy as np
a=np.random.random((10,10))*99+1
##创建一个10*10的单位矩阵B，并将A与B作矩阵乘法，得到矩阵C。打印出矩阵C的值。
b=np.ones((10,10))
c=np.dot(a,b)
print(c)
##条件筛选操作：在矩阵A中找出大于80的元素并替换为0
a[a>80]=0
##分块计算平均值：将10x10矩阵拆分成4个5x5矩阵，求每个子矩阵的均值。
upper,lower=np.split(a,2,axis=0)
block1,block2=np.split(upper,2,axis=1)
block3,block4=np.split(lower,2,axis=1)
mean1=np.mean(block1)
mean2=np.mean(block2)
mean3=np.mean(block3)
mean4=np.mean(block4)
mean=[mean1,mean2,mean3,mean4]
print(mean)
```
